```python
# Classify all columns for disease prediction

def classify_column(col):
    name = col.lower()

    # Target labels and direct diagnostic information
    if any(x in name for x in [
        "disease_label", "disease_severity",
        "diagnosis", "disease_present",
        "endothelium_findings", "endothelium_severity"
    ]):
        return "Target / diagnostic information"

    # Image references and identifiers
    if any(x in name for x in [
        "image_id", "image_file", "filename",
        "image_path", "tissue_serial"
    ]):
        return "Image reference / identifier"

    # Potential image-derived clinical biomarkers
    if any(x in name for x in [
        "ecd", "cv", "hex", "cell_count",
        "cell_area", "density", "morphology",
        "polymegathism", "pleomorphism"
    ]):
        return "Potential biomarker feature"

    # Image quality or acquisition information
    if any(x in name for x in [
        "image_quality", "quality", "focus",
        "resolution", "illumination", "contrast"
    ]):
        return "Potential image-quality feature"

    # Clinical or donor metadata
    if any(x in name for x in [
        "age", "sex", "gender", "donor",
        "death_to", "request_to", "surgery",
        "medical", "clinical", "history"
    ]):
        return "Potential clinical feature"

    # Document and administrative fields
    if any(x in name for x in [
        "pdf", "page", "report", "request",
        "date", "time", "name", "address",
        "review", "comment", "notes"
    ]):
        return "Administrative / requires review"

    return "Unclassified — inspect manually"


column_review = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Missing Count": df.isna().sum().values,
    "Unique Values": df.nunique(dropna=True).values
})

column_review["Initial Category"] = column_review["Column"].apply(
    classify_column
)

display(
    column_review.sort_values(
        ["Initial Category", "Column"]
    ).reset_index(drop=True)
)
```

```python
# Keep only the upper triangle to avoid duplicate pairs
upper_triangle = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

correlation_pairs = (
    upper_triangle.stack()
    .reset_index()
)

correlation_pairs.columns = [
    "Feature_1", "Feature_2", "Correlation"
]

correlation_pairs["Absolute_Correlation"] = (
    correlation_pairs["Correlation"].abs()
)

display(
    correlation_pairs[
        correlation_pairs["Absolute_Correlation"] >= 0.70
    ].sort_values("Absolute_Correlation", ascending=False)
)
```

```python
# Split combined disease labels into individual diseases
disease_lists = (
    df["Disease_Label"]
    .fillna("")
    .astype(str)
    .apply(lambda x: [d.strip() for d in x.split(";") if d.strip()])
)

# Identify all disease categories
all_diseases = sorted({
    disease
    for labels in disease_lists
    for disease in labels
})

print("Disease categories:")
print(all_diseases)
```

```python
disease_targets = pd.DataFrame(index=df.index)

for disease in all_diseases:
    disease_targets[disease] = disease_lists.apply(
        lambda labels: int(disease in labels)
    )

display(disease_targets.sum().sort_values(ascending=False).to_frame("Positive Count"))
```

```python
numeric_df = df.select_dtypes(include=np.number)

feature_target_corr = pd.DataFrame(
    index=numeric_df.columns,
    columns=disease_targets.columns,
    dtype=float
)

for feature in numeric_df.columns:
    for disease in disease_targets.columns:
        valid = numeric_df[feature].notna()

        if valid.sum() > 1 and disease_targets.loc[valid, disease].nunique() > 1:
            feature_target_corr.loc[feature, disease] = (
                numeric_df.loc[valid, feature]
                .corr(disease_targets.loc[valid, disease])
            )

display(feature_target_corr.round(3))
```

```python
plt.figure(figsize=(14, max(6, len(feature_target_corr) * 0.45)))

sns.heatmap(
    feature_target_corr.astype(float),
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1
)

plt.title("Numerical Feature Correlation with Individual Diseases")
plt.xlabel("Disease")
plt.ylabel("Numerical Feature")
plt.tight_layout()
plt.show()
```
