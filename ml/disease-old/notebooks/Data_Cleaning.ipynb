```python
import pandas as pd
import numpy as np
import re
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", 100)
```

```python
excel_path = r"C:\Users\minsa_\Downloads\Combined_Research_Dataset_Final.xlsx"

df_original = pd.read_excel(
    excel_path,
    sheet_name="Combined_Dataset"
)

# Create a separate working copy
df_clean = df_original.copy(deep=True)

print("Dataset loaded successfully")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
```

```python
for i, col in enumerate(df_clean.columns, start=1):
    print(f"{i}. {col}")
```

```python
empty_columns = df_clean.columns[
    df_clean.isna().all()
].tolist()

print("Completely empty columns:", empty_columns)

df_clean = df_clean.drop(
    columns=empty_columns
).copy()

print("\nDataset after removing empty columns")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
```

```python
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns.tolist()

for col in text_columns:
    df_clean[col] = df_clean[col].map(
        lambda x: (
            re.sub(r"\s+", " ", x).strip()
            if isinstance(x, str)
            else x
        )
    )

# Convert empty strings to missing values
for col in text_columns:
    df_clean[col] = df_clean[col].replace("", np.nan)

print("Text formatting cleaned.")
```

```python
normal_mask = (
    df_clean["Disease_Label"]
    .astype("string")
    .str.strip()
    .str.casefold()
    .eq("normal")
)

missing_severity = df_clean["Disease_Severity"].isna()

fill_mask = normal_mask & missing_severity

df_clean.loc[
    fill_mask,
    "Disease_Severity"
] = "None"

print(
    "Normal cases filled with None:",
    fill_mask.sum()
)
```

```python
display(
    df_clean.loc[
        normal_mask,
        ["Disease_Label", "Disease_Severity"]
    ].head(30)
)

print(
    "Remaining missing Disease_Severity values:",
    df_clean["Disease_Severity"].isna().sum()
)
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Exact duplicate rows:", duplicate_count)
```

```python
df_clean["Disease_Label"] = df_clean["Disease_Label"].map(
    lambda x: (
        re.sub(r"\s*;\s*", "; ", x)
        if isinstance(x, str)
        else x
    )
)
```

```python
output_path = r"C:\Users\minsa_\Downloads\Combined_Dataset_Cleaned.xlsx"

df_clean.to_excel(output_path, index=False)

print("Cleaned dataset saved successfully.")
```
