# Component 3 - XAI step 1: why is a region left out of the ROI? (Grad-CAM++)

Grad-CAM++ on the Phase 6 EfficientNet-B0 tile classifier, explaining the **poor** score. Evaluated on the
10 validation reference images (never used for training), rules fixed in `config.yaml -> xai.roi`:
* **faithfulness (primary)** - replacing the top 20 % of the heatmap with neutral tissue lowers P(poor) more than
  replacing the same mask shifted elsewhere in the window;
* **plausibility** - in windows partly labelled poor, the heatmap falls on the poor-labelled part more than its area.

Grad-CAM is reported for comparison. Explanations never change a number of the pipeline.
